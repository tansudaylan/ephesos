# Emitting planetary companions

Ephesos models a planet's contribution to the system light curve and its disappearance behind the star. This example compares secondary eclipses for three planetary surface-brightness ratios.

```python
from pathlib import Path

from IPython.display import Image, display
import numpy as np

from emitting_companion import BRIGHTNESS_RATIOS, run_example

output_path = Path("visuals/emitting_companion.png")
orbital_phase, relative_flux, output_path = run_example(output_path)
eclipse_index = np.argmin(np.abs(orbital_phase - 0.5))
quadrature_index = np.argmin(np.abs(orbital_phase - 0.25))
print(f"Reading from {output_path}...")
display(Image(filename=str(output_path)))
{
    label: {
        "surface_brightness_ratio": BRIGHTNESS_RATIOS[label],
        "quadrature_flux_excess_ppm": 1.0e6 * (flux[quadrature_index] - 1.0),
        "secondary_eclipse_flux_excess_ppm": 1.0e6 * (flux[eclipse_index] - 1.0),
    }
    for label, flux in relative_flux.items()
}
```

## Interpretation

Outside eclipse, the planet adds its own light to the stellar flux. At orbital phase 0.5 the star occults the planet, so each curve returns to the stellar baseline and the eclipse depth directly measures the planet-to-star flux ratio.
